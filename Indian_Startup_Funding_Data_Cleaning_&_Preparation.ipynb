```python
df.columns = ['sr_no', 'date', 'startup_name', 'industry', 'subvertical', 'city', 'investors', 'investment_type', 'amount_usd', 'remarks']
df.head(10)
```

```python
bad = pd.to_datetime(df['date'],format = '%d/%m/%Y', errors = 'coerce').isna()
df.loc[bad,'date']
```

```python
d = df['date'].str.replace(r'\\+xc2\\+xa0', '', regex=True)
d = d.str.replace(r'[./]+', '/', regex=True)
d = d.str.replace(r'^(\d{2}/\d{2})(\d{4})$', r'\1/\2', regex=True)
d = d.str.replace(r'/0(\d{2})$', r'/20\1', regex=True)
df['date'] = pd.to_datetime(d, format='%d/%m/%Y')
df['date'].dtype
```

```python
df.loc[~df['amount_usd'].str.fullmatch(r'[\d,]+', na=False) & df['amount_usd'].notna(), 'amount_usd'].unique()
```

```python
a = df['amount_usd'].str.replace(r'\\+xc2\\+xa0', '', regex=True)
a = a.str.replace(r'[,+]', '', regex=True)
df['amount_usd'] = pd.to_numeric(a, errors='coerce')
df['amount_usd'].dtype
```

```python
df['city'].value_counts().head(40)
```

```python
c= df['city'].str.replace(r'\\+xc2\\+xa0','',regex=True).str.strip()
c= c.str.split(r'\s*/\s*',regex=True).str[0]
c= c.replace({'Bangalore':'Bengaluru', 'Gurgaon': 'Gurugram', 'Delhi': 'New Delhi', 'Ahemadabad':'Ahmedabad', 'Trivandrum':'Thiruvananthapuram'})

df['city']=c.fillna('Unknown')
df['city'].value_counts().head(15)
```

```python
df['investment_type'].value_counts()
```

```python
def fix_type(x):
    if pd.isna(x):
        return 'Unknown'
    t = x.lower()
    if 'pre' in t and 'series a' in t:
        return 'Pre-Series A'
    for L in 'abcdefghij':
        if 'series ' + L in t:
            return 'Series ' + L.upper()
    if 'seed' in t or 'angel' in t or 'angle' in t:
        return 'Seed/Angel'
    if 'debt' in t or 'loan' in t or 'mezzanine' in t:
        return 'Debt'
    if 'private' in t or 'equity' in t:
        return 'Private Equity'
    if 'crowd' in t:
        return 'Crowdfunding'
    return 'Other'

df['investment_type'] = df['investment_type'].apply(fix_type)
df['investment_type'].value_counts()
```

```python
print(df['industry'].nunique())
df['industry'].value_counts().head(50)
```

```python
k = df['industry'].str.lower().str.replace(r'[^a-z]', '', regex=True)

mapping = {'ecommerce':'eCommerce', 'edtech':'EdTech', 'fintech':'FinTech',
           'foodbeverage':'Food & Beverage', 'foodandbeverage':'Food & Beverage',
           'foodbeverages':'Food & Beverage', 'foodandbeverages':'Food & Beverage',
           'it':'Technology', 'informationtechnology':'Technology',
           'transport':'Transportation', 'others':'Other'}

df['industry'] = k.map(mapping).fillna(df['industry']).fillna('Unknown')
df['industry'].nunique()
```

```python
import re

def decode_junk(m):
    hexes = re.findall(r'x([0-9a-fA-F]{2})', m.group())
    return bytes(int(h, 16) for h in hexes).decode('utf-8', errors='ignore')

for col in ['startup_name', 'subvertical', 'investors', 'remarks']:
    s = df[col].str.replace(r'(?:\\+x[0-9a-fA-F]{2})+', decode_junk, regex=True)
    s = s.str.replace(r'\\+n', ' ', regex=True)
    df[col] = s.str.replace(r'\s+', ' ', regex=True).str.strip()

print(df['startup_name'].str.contains(r'\\+x|\\+n', regex=True).sum())
print(df['investors'].str.contains(r'\\+x|\\+n', regex=True, na=False).sum())
```
